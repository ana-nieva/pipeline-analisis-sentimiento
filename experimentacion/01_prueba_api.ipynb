```python
import requests
import pandas as pd

# url de la api apuntando al endpoint de reseñas
url = "https://amazon-reviews-api-g5ae.onrender.com/reviews"

print("conectandose a la api...")
res = requests.get(url, params={"page": 1})

print("código de estado:", res.status_code)

datos = res.json()
print("tipo:", type(datos))

# si los datos vienen envueltos en un diccionario (por la paginación), buscamos la clave
if type(datos) == dict:
    print("claves disponibles:", datos.keys())
    # generalmente los registros vienen en la clave 'data' o 'reviews'
    if "data" in datos:
        df = pd.DataFrame(datos["data"])
        display(df.head())
else:
    df = pd.DataFrame(datos)
    display(df.head())
```
